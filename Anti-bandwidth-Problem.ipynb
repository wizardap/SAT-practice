## Encoding AMO constraints in each block

$$
\bigwedge_{j=1}^{w-1} x_{i,j} \rightarrow R_{i,j} \tag{1}
$$

$$
\bigwedge_{j=2}^{w-1} R_{i,j-1} \rightarrow R_{i,j} \tag{2}
$$

$$
\neg x_{i,1} \rightarrow \neg R_{i,1} \tag{3}
$$

$$
\bigwedge_{j=2}^{w-1} \neg x_{i,j} \land \neg R_{i,j-1} \rightarrow \neg R_{i,j} \tag{4}
$$

$$
\bigwedge_{j=2}^{w} x_{i,j} \rightarrow \neg R_{i,j-1} \tag{5}
$$


## Clauses connecting two adjacent blocks

$$
\begin{aligned}

& \bigwedge_{j=2}^{w} \left( \sum_{q=1}^{w-j+1} x_{i,q} \leq 0 \lor \sum_{q'=1}^{j-1} x_{i+1,q'} \leq 0 \right) \\
\iff & \bigwedge_{j=2}^{w} \left( R_{i,w-j+1} \lor R_{i+1,j-1} \right) 
\end{aligned}
$$
