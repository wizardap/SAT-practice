# NSC Encoding

Gọi $R_{i,j}$ là biến phụ biểu diễn trạng thái:

$$
R_{i,j}=TRUE \Rightarrow \sum_{t=1}^{i}X_t \ge j
$$

Nói cách khác, sau khi xét $i$ biến đầu tiên, bộ đếm đã đạt ít nhất $j$ giá trị `TRUE`.

## 1. Khởi tạo bộ đếm

$$
\bigwedge_{i=1}^{n-1}(X_i \rightarrow R_{i,1})
$$

Nếu $X_i=TRUE$ thì trạng thái “đã có ít nhất 1 biến TRUE” tại vị trí $i$ phải được bật.

## 2. Lan truyền trạng thái TRUE

$$
\bigwedge_{i=2}^{n-1}\;\bigwedge_{j=1}^{\min(i-1,k)}
(R_{i-1,j} \rightarrow R_{i,j})
$$

Nếu prefix trước đã đạt mức $j$, thì prefix hiện tại cũng vẫn đạt mức $j$.

## 3. Tăng bộ đếm khi gặp biến TRUE

$$
\bigwedge_{i=2}^{n-1}\;\bigwedge_{j=2}^{\min(i,k)}
(X_i \land R_{i-1,j-1} \rightarrow R_{i,j})
$$

Nếu $X_i=TRUE$ và trước đó đã có ít nhất $j-1$ biến TRUE, thì hiện tại đạt ít nhất $j$ biến TRUE.

## 4. Giữ trạng thái FALSE

$$
\bigwedge_{i=2}^{n-1}\;\bigwedge_{j=1}^{\min(i-1,k)}
(\neg X_i \land \neg R_{i-1,j} \rightarrow \neg R_{i,j})
$$

Nếu $X_i=FALSE$ và prefix trước chưa đạt mức $j$, thì prefix hiện tại cũng chưa thể đạt mức $j$.

## 5. Chặn trạng thái đường chéo

$$
\bigwedge_{i=1}^{k}(\neg X_i \rightarrow \neg R_{i,i})
$$

Nếu $X_i=FALSE$ thì trạng thái mới mở rộng $R_{i,i}$ cũng phải là `FALSE`.

## 6. Ngăn nhảy bậc

$$
\bigwedge_{i=2}^{n-1}\;\bigwedge_{j=2}^{\min(i,k)}
(\neg R_{i-1,j-1} \rightarrow \neg R_{i,j})
$$

Nếu trước đó chưa đạt $j-1$ biến TRUE, thì hiện tại không thể tự nhảy lên mức $j$.

## 7. Điều kiện cuối cho NSC $\ge k$

Với $1 \le k < n$:

$$
R_{n-1,k} \lor (X_n \land R_{n-1,k-1})
$$

Điều kiện này đảm bảo toàn bộ dãy có ít nhất $k$ biến TRUE.

## 8. Điều kiện cuối cho NSC $\le k$

$$
\bigwedge_{i=k+1}^{n}(X_i \rightarrow \neg R_{i-1,k})
$$

Nếu prefix trước đã đạt đủ $k$ biến TRUE, thì biến tiếp theo không được TRUE nữa. Đây là phần chặn để tổng không vượt quá $k$.

# AMK - At Most K

Ràng buộc AMK yêu cầu trong $n$ biến Boolean có **không quá $k$ biến TRUE**:

$$
\sum_{i=1}^{n} X_i \le k
$$

Với NSC, AMK được tạo từ 4 nhóm công thức:

| Nhóm | Vai trò |
| :--- | :--- |
| (1) | Khởi tạo bộ đếm khi gặp biến TRUE đầu tiên |
| (2) | Lan truyền trạng thái đã đạt mức đếm $j$ |
| (3) | Tăng bộ đếm khi $X_i=TRUE$ |
| (8) | Chặn không cho số lượng TRUE vượt quá $k$ |

Trong đó, các công thức (1), (2), (3) dùng để xây dựng sequential counter. Công thức (8) là điều kiện kiểm tra cuối: nếu prefix trước đã đạt $k$ biến TRUE thì biến tiếp theo không được TRUE nữa.

Do đó:

$$
NSC_{\le k} = (1) \land (2) \land (3) \land (8)
$$

# ALK - At Least K

Ràng buộc ALK yêu cầu trong $n$ biến Boolean có **ít nhất $k$ biến TRUE**:

$$
\sum_{i=1}^{n} X_i \ge k
$$

Với NSC, ALK cần nhiều công thức hơn AMK vì phải đảm bảo bộ đếm phản ánh đúng cả hai chiều: trạng thái TRUE phải được bật khi đủ điều kiện, và trạng thái FALSE phải được giữ khi chưa đủ điều kiện.

| Nhóm | Vai trò |
| :--- | :--- |
| (1) | Khởi tạo bộ đếm khi gặp biến TRUE đầu tiên |
| (2) | Lan truyền trạng thái đã đạt mức đếm $j$ |
| (3) | Tăng bộ đếm khi $X_i=TRUE$ |
| (4) | Giữ trạng thái chưa đạt mức $j$ khi $X_i=FALSE$ |
| (5) | Chặn trạng thái đường chéo không hợp lệ |
| (6) | Ngăn bộ đếm nhảy từ mức thấp lên mức cao không có cơ sở |
| (7) | Điều kiện cuối để ép toàn bộ dãy đạt ít nhất $k$ biến TRUE |

Các công thức (1) đến (6) xây dựng bộ đếm tuần tự. Công thức (7) là điều kiện kiểm tra cuối: sau khi xét toàn bộ dãy, bộ đếm phải chứng minh đã có ít nhất $k$ biến TRUE.

Do đó:

$$
NSC_{\ge k} = (1) \land (2) \land (3) \land (4) \land (5) \land (6) \land (7)
$$

# EK - Exactly K

Ràng buộc EK yêu cầu trong $n$ biến Boolean có **đúng $k$ biến TRUE**:

$$
\sum_{i=1}^{n} X_i = k
$$

Ý tưởng là kết hợp hai ràng buộc đã có:

| Ràng buộc | Vai trò |
| :--- | :--- |
| $NSC_{\ge k}$ | Đảm bảo có ít nhất $k$ biến TRUE |
| $NSC_{\le k}$ | Đảm bảo có nhiều nhất $k$ biến TRUE |

Khi cả hai điều kiện cùng đúng, ta có:

$$
\sum_{i=1}^{n} X_i \ge k
\quad \land \quad
\sum_{i=1}^{n} X_i \le k
\quad \Rightarrow \quad
\sum_{i=1}^{n} X_i = k
$$

Vì vậy, EK dùng các công thức của cả ALK và AMK:

- (1), (2), (3), (4), (5), (6), (7) để tạo $NSC_{\ge k}$.
- (1), (2), (3), (8) để tạo $NSC_{\le k}$.

Kết luận:

$$
NSC_{=k} = NSC_{\ge k} \land NSC_{\le k}
$$


# Range Encoding - $NSC_{[u,v]}$

Range encoding yêu cầu số biến TRUE nằm trong khoảng từ $u$ đến $v$:

$$
u \le \sum_{i=1}^{n} X_i \le v
$$

Ý tưởng là kết hợp hai ràng buộc:

| Ràng buộc | Vai trò |
| :--- | :--- |
| $NSC_{\ge u}$ | Đảm bảo có ít nhất $u$ biến TRUE |
| $NSC_{\le v}$ | Đảm bảo có nhiều nhất $v$ biến TRUE |

Khi cả hai điều kiện cùng đúng, tổng số biến TRUE bị kẹp trong khoảng $[u,v]$:

$$
\sum_{i=1}^{n} X_i \ge u
\quad \land \quad
\sum_{i=1}^{n} X_i \le v
\quad \Rightarrow \quad
u \le \sum_{i=1}^{n} X_i \le v
$$

Các nhóm công thức được dùng như sau:

| Phần | Công thức sử dụng | Tham số |
| :--- | :--- | :--- |
| Cận dưới | (1), (2), (3), (4), (5), (6), (7) | $u$ |
| Cận trên | (1), (2), (3), (8) | $v$ |

Kết luận:

$$
NSC_{[u,v]} = NSC_{\ge u} \land NSC_{\le v}
$$

## Code mẫu

```python
class NSCEncoding:
    """Normalized Sequential Counter encoder for cardinality constraints.

    Thuộc tính chính:
    - self.clauses: danh sách CNF clauses được sinh ra.
    - self.top: id biến lớn nhất hiện đang được dùng.
    """

    def __init__(self, top=0):
        # top nên được khởi tạo bằng id lớn nhất của biến gốc.
        # Ví dụ variables = [1, 2, 3, 4] thì top = 4.
        self.top = top
        self.clauses = []

    def new_var(self):
        # Mỗi lần cần biến phụ R_{i,j}, tăng top lên 1 để lấy id mới.
        self.top += 1
        return self.top

    def new_counter_states(self, variables, k):
        """Create R[(i, j)] for prefix variables[0..i] reaching at least j + 1 TRUEs."""
        R = {}

        # Không tạo state cho biến cuối cùng. Biến cuối được xử lý riêng
        # trong điều kiện cuối của AMK/ALK.
        for i in range(len(variables) - 1):
            for j in range(min(i + 1, k)):
                R[(i, j)] = self.new_var()
        return R

    def amk(self, variables, k):
        """Encode sum(variables) <= k using NSC formulas (1), (2), (3), (8)."""
        n = len(variables)

        if k < 0:
            # At most số âm là bất khả thi, thêm empty clause để UNSAT.
            self.clauses.append([])
            # Nếu k >= n thì ràng buộc luôn đúng, không cần thêm clause.
            return {}
        if k >= n:
            return {}
        if k == 0:
            # At most 0 nghĩa là mọi biến đều phải FALSE.
            self.clauses.extend([-x] for x in variables)
            return {}

        R = self.new_counter_states(variables, k)

        # (1) X_i -> R_i,1: nếu X_i TRUE thì prefix tới i có >= 1 TRUE.
        for i in range(n - 1):
            self.clauses.append([-variables[i], R[(i, 0)]])

        # (2) R_{i-1,j} -> R_{i,j}: trạng thái đã đạt mức j được lan truyền.
        for i in range(1, n - 1):
            for j in range(min(i, k)):
                self.clauses.append([-R[(i - 1, j)], R[(i, j)]])

        # (3) X_i and R_{i-1,j-1} -> R_{i,j}: gặp TRUE thì tăng bộ đếm.
        for i in range(1, n - 1):
            for j in range(1, min(i + 1, k)):
                self.clauses.append([-variables[i], -R[(i - 1, j - 1)], R[(i, j)]])

        # (8) Nếu prefix trước đã đạt k TRUE thì biến tiếp theo không được TRUE.
        for i in range(k, n):
            self.clauses.append([-variables[i], -R[(i - 1, k - 1)]])

        return R

    def alk(self, variables, k):
        """Encode sum(variables) >= k using NSC formulas (1)..(7)."""
        n = len(variables)

        if k <= 0:
            # At least <= 0 luôn đúng.
            return {}
        if k > n:
            # At least nhiều hơn n là bất khả thi.
            self.clauses.append([])
            return {}
        if k == n:
            # At least n nghĩa là mọi biến đều phải TRUE.
            self.clauses.extend([x] for x in variables)
            return {}

        R = self.new_counter_states(variables, k)

        # (1) X_i -> R_i,1.
        for i in range(n - 1):
            self.clauses.append([-variables[i], R[(i, 0)]])

        # (2) Lan truyền trạng thái TRUE.
        for i in range(1, n - 1):
            for j in range(min(i, k)):
                self.clauses.append([-R[(i - 1, j)], R[(i, j)]])

        # (3) Tăng bộ đếm khi X_i TRUE.
        for i in range(1, n - 1):
            for j in range(1, min(i + 1, k)):
                self.clauses.append([-variables[i], -R[(i - 1, j - 1)], R[(i, j)]])

        # (4) Nếu X_i FALSE và trước đó chưa đạt j, hiện tại cũng chưa đạt j.
        for i in range(1, n - 1):
            for j in range(min(i, k)):
                self.clauses.append([variables[i], R[(i - 1, j)], -R[(i, j)]])

        # (5) Chặn trạng thái đường chéo R_{i,i} khi X_i FALSE.
        for i in range(min(k, n - 1)):
            self.clauses.append([variables[i], -R[(i, i)]])

        # (6) Ngăn bộ đếm nhảy bậc không có cơ sở.
        for i in range(1, n - 1):
            for j in range(1, min(i + 1, k)):
                self.clauses.append([R[(i - 1, j - 1)], -R[(i, j)]])

        # (7) Điều kiện cuối: toàn bộ dãy phải có ít nhất k biến TRUE.
        last_x = variables[-1]
        last_prefix = n - 2
        if k == 1:
            self.clauses.append([R[(last_prefix, 0)], last_x])
        else:
            self.clauses.append([R[(last_prefix, k - 1)], last_x])
            self.clauses.append([R[(last_prefix, k - 1)], R[(last_prefix, k - 2)]])

        return R

    def exk(self, variables, k):
        """Encode sum(variables) == k bằng cách ghép ALK và AMK."""
        return {
            "at_least": self.alk(variables, k),
            "at_most": self.amk(variables, k),
        }

    def range(self, variables, u, v):
        """Encode u <= sum(variables) <= v bằng cách ghép cận dưới và cận trên."""
        return {
            "lower": self.alk(variables, u),
            "upper": self.amk(variables, v),
        }


# Giữ alias này để các cell cũ dùng NSCEncoder vẫn chạy được.
NSCEncoder = NSCEncoding
```

### Bài toán N quân hậu

Ta biểu diễn mỗi ô bàn cờ bằng một biến Boolean:

$$
q_{r,c}=TRUE \iff \text{có quân hậu ở hàng } r \text{, cột } c
$$

Với bàn cờ $n \times n$, bài toán cần các ràng buộc:

| Nhóm ô | Ràng buộc | Cách mã hóa |
| :--- | :--- | :--- |
| Mỗi hàng | Đúng 1 quân hậu | `exk(row_vars, 1)` |
| Mỗi cột | Đúng 1 quân hậu | `exk(col_vars, 1)` |
| Mỗi đường chéo chính | Nhiều nhất 1 quân hậu | `amk(diagonal_vars, 1)` |
| Mỗi đường chéo phụ | Nhiều nhất 1 quân hậu | `amk(diagonal_vars, 1)` |

Sau khi sinh CNF bằng `NSCEncoding`, ta đưa toàn bộ clauses vào `Glucose4` để tìm nghiệm.

```python
from pysat.solvers import Glucose4


def queen_var(row, col, n):
    """Map ô (row, col) sang id biến CNF, bắt đầu từ 1."""
    return row * n + col + 1


def build_n_queens_cnf(n):
    """Sinh CNF cho bài toán N quân hậu bằng NSCEncoding."""
    encoder = NSCEncoding(top=n * n)

    # Mỗi hàng có đúng 1 quân hậu.
    for row in range(n):
        row_vars = [queen_var(row, col, n) for col in range(n)]
        encoder.exk(row_vars, 1)

    # Mỗi cột có đúng 1 quân hậu.
    for col in range(n):
        col_vars = [queen_var(row, col, n) for row in range(n)]
        encoder.exk(col_vars, 1)

    # Đường chéo chính: row - col là hằng số.
    for diff in range(-(n - 1), n):
        diagonal = [
            queen_var(row, row - diff, n)
            for row in range(n)
            if 0 <= row - diff < n
        ]
        if len(diagonal) > 1:
            encoder.amk(diagonal, 1)

    # Đường chéo phụ: row + col là hằng số.
    for total in range(2 * n - 1):
        diagonal = [
            queen_var(row, total - row, n)
            for row in range(n)
            if 0 <= total - row < n
        ]
        if len(diagonal) > 1:
            encoder.amk(diagonal, 1)

    return encoder.clauses, encoder.top


def solve_n_queens(n):
    clauses, top = build_n_queens_cnf(n)

    with Glucose4(bootstrap_with=clauses) as solver:
        if not solver.solve():
            return None, clauses, top

        model = {lit for lit in solver.get_model() if lit > 0}

    board = []
    for row in range(n):
        board.append([
            "Q" if queen_var(row, col, n) in model else "."
            for col in range(n)
        ])

    return board, clauses, top


n = 8
board, clauses, top = solve_n_queens(n)

if board is None:
    print("UNSAT")
else:
    print(f"clauses: {len(clauses)}, variables: {top}")
    print("\n".join(" ".join(row) for row in board))
```

### Bài toán Sudoku

Ta biểu diễn mỗi khả năng điền số bằng một biến Boolean:

$$
x_{r,c,d}=TRUE \iff \text{ô }(r,c)\text{ chứa số }d
$$

Với Sudoku chuẩn $9 \times 9$, các ràng buộc chính đều là dạng **đúng một**:

| Nhóm biến | Ý nghĩa | Cách mã hóa |
| :--- | :--- | :--- |
| Một ô $(r,c)$ với 9 chữ số | Mỗi ô nhận đúng 1 số | `exk(cell_vars, 1)` |
| Một hàng $r$ với số $d$ | Mỗi số xuất hiện đúng 1 lần trên hàng | `exk(row_digit_vars, 1)` |
| Một cột $c$ với số $d$ | Mỗi số xuất hiện đúng 1 lần trên cột | `exk(col_digit_vars, 1)` |
| Một khối $3 \times 3$ với số $d$ | Mỗi số xuất hiện đúng 1 lần trong khối | `exk(box_digit_vars, 1)` |

Các số đã cho trong đề bài được thêm bằng unit clause `[x_{r,c,d}]`. Sau đó dùng `Glucose4` để tìm model thỏa mãn toàn bộ CNF.

```python
from pysat.solvers import Glucose4


def sudoku_var(row, col, digit):
    """Map (row, col, digit) sang CNF variable id.

    row, col: 0..8
    digit: 1..9
    """
    return row * 81 + col * 9 + digit


def build_sudoku_cnf(grid):
    """Sinh CNF cho Sudoku 9x9 bằng NSCEncoding."""
    encoder = NSCEncoding(top=9 * 9 * 9)

    # Mỗi ô có đúng 1 chữ số.
    for row in range(9):
        for col in range(9):
            cell_vars = [sudoku_var(row, col, digit) for digit in range(1, 10)]
            encoder.exk(cell_vars, 1)

    # Mỗi hàng chứa mỗi chữ số đúng 1 lần.
    for row in range(9):
        for digit in range(1, 10):
            row_digit_vars = [sudoku_var(row, col, digit) for col in range(9)]
            encoder.exk(row_digit_vars, 1)

    # Mỗi cột chứa mỗi chữ số đúng 1 lần.
    for col in range(9):
        for digit in range(1, 10):
            col_digit_vars = [sudoku_var(row, col, digit) for row in range(9)]
            encoder.exk(col_digit_vars, 1)

    # Mỗi khối 3x3 chứa mỗi chữ số đúng 1 lần.
    for box_row in range(0, 9, 3):
        for box_col in range(0, 9, 3):
            for digit in range(1, 10):
                box_digit_vars = [
                    sudoku_var(row, col, digit)
                    for row in range(box_row, box_row + 3)
                    for col in range(box_col, box_col + 3)
                ]
                encoder.exk(box_digit_vars, 1)

    # Các ô đã cho trong đề bài.
    for row in range(9):
        for col in range(9):
            digit = grid[row][col]
            if digit != 0:
                encoder.clauses.append([sudoku_var(row, col, digit)])

    return encoder.clauses, encoder.top


def solve_sudoku(grid):
    clauses, top = build_sudoku_cnf(grid)

    with Glucose4(bootstrap_with=clauses) as solver:
        if not solver.solve():
            return None, clauses, top

        model = {lit for lit in solver.get_model() if lit > 0}

    solution = [[0 for _ in range(9)] for _ in range(9)]
    for row in range(9):
        for col in range(9):
            for digit in range(1, 10):
                if sudoku_var(row, col, digit) in model:
                    solution[row][col] = digit
                    break

    return solution, clauses, top


puzzle = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9],
]

solution, clauses, top = solve_sudoku(puzzle)

if solution is None:
    print("UNSAT")
else:
    print(f"clauses: {len(clauses)}, variables: {top}")
    for row in solution:
        print(" ".join(str(value) for value in row))
```
